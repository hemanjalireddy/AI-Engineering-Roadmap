```python
from vllm import LLM, SamplingParams
import torch
```

```python
if not torch.cuda.is_available():
    print("CUDA is not available. Please check your Colab runtime settings.")
else:
    print("CUDA is available. Proceeding with vLLM initialization.")
    llm = LLM("gpt2")
```

### I am trying to make AI write a thriller movie plot lol

```python
prompt = "There is a man standing by the window pane. In his hand he holds a"

sampling_params = SamplingParams(max_tokens=100, temperature=0.7, top_p=0.9)
```

```python
completion = llm.generate([prompt], sampling_params)

print(completion[0].outputs[0].text)
```

I think it gave an okay story. It was giving me bizzare story plots when the temparature was high and I wonder why?
